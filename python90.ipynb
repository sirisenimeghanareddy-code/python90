## 13. Percentage and Grade

Write a Python program to input marks in five subjects: Physics, Chemistry, Biology, Mathematics, and Computer. Calculate the percentage and grade according to the following criteria:

- Percentage >= 90%: Grade A
- Percentage >= 80%: Grade B
- Percentage >= 70%: Grade C
- Percentage >= 60%: Grade D
- Percentage >= 40%: Grade E
- Percentage < 40%: Grade F

```python
marks = []

for subject in ["Physics", "Chemistry", "Biology", "Mathematics", "Computer"]:
    marks.append(float(input(f"Enter {subject} marks: ")))

percentage = sum(marks) / 5

if percentage >= 90:
    grade = "A"
elif percentage >= 80:
    grade = "B"
elif percentage >= 70:
    grade = "C"
elif percentage >= 60:
    grade = "D"
elif percentage >= 40:
    grade = "E"
else:
    grade = "F"

print("Percentage =", percentage, "%")
print("Grade =", grade)
```

## 14. Gross Salary Calculation

Write a Python program to input the basic salary of an employee and calculate the gross salary according to the following conditions:

- Basic Salary <= 10,000: HRA = 20%, DA = 80%
- Basic Salary <= 20,000: HRA = 25%, DA = 90%
- Basic Salary > 20,000: HRA = 30%, DA = 95%

```python
salary = float(input("Enter basic salary: "))

if salary <= 10000:
    hra, da = 0.20, 0.80
elif salary <= 20000:
    hra, da = 0.25, 0.90
else:
    hra, da = 0.30, 0.95

gross = salary + salary * hra + salary * da

print("Gross salary =", gross)
```

## 15. Electricity Bill Calculation

Write a Python program to input electricity units and calculate the total electricity bill according to the following conditions:

- First 50 units: Rs. 0.50 per unit
- Next 100 units: Rs. 0.75 per unit
- Next 100 units: Rs. 1.20 per unit
- Above 250 units: Rs. 1.50 per unit
- Add an additional surcharge of 20% to the bill.

```python
units = float(input("Enter electricity units: "))

if units <= 50:
    bill = units * 0.50
elif units <= 150:
    bill = 50 * 0.50 + (units - 50) * 0.75
elif units <= 250:
    bill = 50 * 0.50 + 100 * 0.75 + (units - 150) * 1.20
else:
    bill = 50 * 0.50 + 100 * 0.75 + 100 * 1.20 + (units - 250) * 1.50

bill += bill * 0.20

print("Total electricity bill =", bill)
```

42. Find All Prime Factors of a Number

```python
n = int(input("Enter a number: "))
num = abs(n)
i = 2

print("Prime factors:")

while i * i <= num:
    if num % i == 0:
        print(i, end=" ")
        while num % i == 0:
            num //= i
    i += 1

if num > 1:
    print(num, end=" ")
```

43. Check Whether a Number Is an Armstrong Number

Write a Python program to check whether a number is an Armstrong number or not.

```python
n = int(input("Enter a number: "))
digits = str(n)
power = len(digits)

total = sum(int(digit) ** power for digit in digits)

if n >= 0 and total == n:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

79. Separate Even, Odd and Prime Numbers into Files

Question: Write a Python program to read numbers from a file and store even, odd, and prime numbers in separate files.

```python
with open("numbers.txt", "w") as f:
    f.write("2 3 4 5 6 7 8 9 10")

with open("numbers.txt", "r") as f:
    nums = list(map(int, f.read().split()))

even = [n for n in nums if n % 2 == 0]
odd = [n for n in nums if n % 2 != 0]

prime = []
for n in nums:
    if n > 1:
        for i in range(2, n):
            if n % i == 0:
                break
        else:
            prime.append(n)

for name, values in [("even.txt", even),
                     ("odd.txt", odd),
                     ("prime.txt", prime)]:
    with open(name, "w") as f:
        f.write(" ".join(map(str, values)))

print("Numbers separated into files")
```

```python
with open("sample.txt", "w") as f:
    f.write("Hello Python\nWelcome")

with open("sample.txt", "r") as f:
    data = f.read()

print("Characters =", len(data))
print("Words =", len(data.split()))
print("Lines =", len(data.splitlines()))
```

83. Check Prime, Armstrong and Perfect Numbers Using Functions

Question: Write a Python program to check whether a number is prime, Armstrong, and perfect using functions.

```python
def prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

def armstrong(n):
    digits = str(n)
    return n >= 0 and sum(int(d) ** len(digits) for d in digits) == n

def perfect(n):
    return n > 0 and sum(i for i in range(1, n)
                         if n % i == 0) == n

n = int(input("Enter a number: "))

print("Prime:", prime(n))
print("Armstrong:", armstrong(n))
print("Perfect:", perfect(n))
```

84. Print Prime Numbers in an Interval Using a Function

Question: Write a Python program to print prime numbers in a given interval using a function.

```python
def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

start = int(input("Enter start: "))
end = int(input("Enter end: "))

for n in range(start, end + 1):
    if is_prime(n):
        print(n, end=" ")
```

85. Print Strong Numbers in an Interval

Question: Write a Python program to print strong numbers in a given interval.

```python
import math

start = int(input("Enter start: "))
end = int(input("Enter end: "))

for n in range(start, end + 1):
    total = sum(math.factorial(int(d)) for d in str(n))
    if n == total:
        print(n, end=" ")
```

86. Print Armstrong Numbers in an Interval

Question: Write a Python program to print Armstrong numbers in a given interval.

```python
start = int(input("Enter start: "))
end = int(input("Enter end: "))

for n in range(start, end + 1):
    digits = str(n)
    total = sum(int(d) ** len(digits) for d in digits)

    if n == total:
        print(n, end=" ")
```

87. Print Perfect Numbers in an Interval

Question: Write a Python program to print perfect numbers in a given interval.

```python
start = int(input("Enter start: "))
end = int(input("Enter end: "))

for n in range(start, end + 1):
    total = sum(i for i in range(1, n) if n % i == 0)

    if n > 0 and total == n:
        print(n, end=" ")
```

89. ATM Transaction

Question: Write a Python program to perform an ATM transaction by checking the balance and withdrawing money.

```python
balance = 5000
amount = int(input("Enter withdrawal amount: "))

if amount <= 0:
    print("Invalid amount")
elif amount <= balance:
    balance -= amount
    print("Withdrawal successful")
    print("Remaining balance =", balance)
else:
    print("Insufficient balance")
```
